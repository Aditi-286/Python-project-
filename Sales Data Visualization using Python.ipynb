```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Random Seed
np.random.seed(10)

# Total Records
records = 50000

# Gender Column
# 40% Men and 60% Women
gender = np.random.choice(
    ['Men', 'Women'],
    records,
    p=[0.4, 0.6]
)

# Product Categories
category = np.random.choice(
    ['Electronics', 'Fashion', 'Groceries', 'Beauty', 'Sports'],
    records
)

# Different Sales for Men and Women
sales = []

for g in gender:

    if g == 'Men':
        amount = np.random.randint(500, 7000)

    else:
        amount = np.random.randint(3000, 12000)

    sales.append(amount)

# Convert to NumPy Array
sales = np.array(sales)

# Quantity Purchased
quantity = np.random.randint(
    1,
    10,
    records
)

# Months
months = np.random.choice(
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    records
)
```

```python
df = pd.DataFrame({
    'Gender': gender,
    'Category': category,
    'Sales': sales,
    'Quantity': quantity,
    'Month': months
})
```

```python
print("FIRST 5 RECORDS")

print(df.head())
```

```python
print(df.info())
```

```python
gender_sales = df.groupby('Gender')['Sales'].sum()

print("\nTOTAL SALES BY GENDER")

print(gender_sales)
```

```python
plt.figure(figsize=(7,5))

gender_sales.plot(kind='bar')

plt.title("Total Sales by Gender")

plt.xlabel("Gender")

plt.ylabel("Sales Amount")

plt.show()
```

```python
plt.figure(figsize=(6,6))

gender_sales.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Sales Percentage by Gender")

plt.ylabel("")

plt.show()
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()

# Correct Month Order
month_order = [
    'Jan', 'Feb', 'Mar', 'Apr',
    'May', 'Jun', 'Jul', 'Aug',
    'Sep', 'Oct', 'Nov', 'Dec'
]

monthly_sales = monthly_sales.reindex(month_order)

print(monthly_sales)
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

records = 50000

gender = np.random.choice(['Men', 'Women'], records, p=[0.4, 0.6])

category = np.random.choice(
    ['Electronics', 'Fashion', 'Groceries', 'Beauty', 'Sports'],
    records,
    p=[0.25, 0.30, 0.15, 0.20, 0.10]   # realistic distribution
)

sales = []

for cat in category:

    if cat == "Electronics":
        amount = np.random.randint(5000, 20000)

    elif cat == "Fashion":
        amount = np.random.randint(3000, 12000)

    elif cat == "Beauty":
        amount = np.random.randint(2000, 9000)

    elif cat == "Sports":
        amount = np.random.randint(1500, 8000)

    else:  # Groceries
        amount = np.random.randint(500, 4000)

    sales.append(amount)

sales = np.array(sales)

df = pd.DataFrame({
    "Gender": gender,
    "Category": category,
    "Sales": sales
})

print(df.head())
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print("CATEGORY WISE SALES:\n")
print(category_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Category Wise Sales (Realistic Distribution)")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```
